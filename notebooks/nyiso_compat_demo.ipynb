# NYISOToolkit Compatibility Demo

This notebook demonstrates how to use NYISOToolkit's analytics and visualization
tools with any ISO through our compatibility layer.

## Features:
- Use NYISOStat-style analytics for any ISO
- Use NYISOVis-style visualizations for any ISO
- Compare multiple ISOs side-by-side


```python
import sys
from pathlib import Path

# Add parent directory to path
sys.path.insert(0, str(Path().resolve().parent))

import pandas as pd
import matplotlib.pyplot as plt
from analytics.iso_stat import ISOStat, compare_isos
from visualization.iso_vis import ISOVis, compare_isos_visualization

# Set style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (14, 6)
```

## 1. NYISO Statistics (Using Compatibility Layer)


```python
# Create statistics object for NYISO
nyiso_stat = ISOStat(iso_code='NYISO', year=2024)

# Get price statistics
price_stats = nyiso_stat.get_price_statistics('lbmp_dam_h')
print("NYISO Price Statistics:")
print(price_stats)
```

## 2. NYISO Visualizations


```python
# Create visualization object
nyiso_vis = ISOVis(iso_code='NYISO', year=2024)

# Generate price PDF
fig = nyiso_vis.fig_price_pdf(dataset='lbmp_dam_h', zones=['NYC', 'LI', 'WEST'])
plt.show()
```

```python
# Generate price time series
fig = nyiso_vis.fig_price_timeseries(
    dataset='lbmp_dam_h',
    zones=['NYC', 'LI'],
    start_date='2024-01-01',
    end_date='2024-01-07'
)
plt.show()
```

## 3. Average Daily Load Pattern


```python
# Get average daily load pattern
avg_load = nyiso_stat.table_average_day_load()

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(avg_load.index, avg_load['Load'], linewidth=2, marker='o')
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Average Load (GWh)')
ax.set_title('NYISO Average Daily Load Pattern - 2024')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Zone Comparison


```python
# Box plot comparing zones
fig = nyiso_vis.fig_zone_comparison(
    dataset='lbmp_dam_h',
    zones=['NYC', 'LI', 'WEST', 'CENTRL', 'NORTH']
)
plt.show()
```
